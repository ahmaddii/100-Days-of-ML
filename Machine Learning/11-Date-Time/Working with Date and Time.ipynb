```python
date["date_dow_name"] = date["date"].dt.day_name()
```

```python
date.drop(columns=["product_id","city_id","orders"]).head()
```

```python
# is weekend?

date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
date['quarter'] = date['date'].dt.quarter # extract the quarter 

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2) # extracf the semester 

date.drop(columns=['product_id','city_id','orders']).head()
```
