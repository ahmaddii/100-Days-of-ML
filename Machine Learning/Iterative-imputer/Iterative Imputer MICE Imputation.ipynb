```python
df_current = df1.copy()

missing_positions = [
    (1, 0),
    (3, 1),
    (49, 2)
]

for iteration in range(1, 101):

    df_previous = df_current.copy()

    # ---- R&D Spend (row 1) ----
    X = df_current.iloc[[0, 2, 3, 4], [1, 2]]
    y = df_current.iloc[[0, 2, 3, 4], 0]

    lr = LinearRegression()
    lr.fit(X, y)

    df_current.iloc[1, 0] = lr.predict(
        df_current.iloc[1, [1, 2]].values.reshape(1, 2)
    )[0]

    # ---- Administration (row 3) ----
    X = df_current.iloc[[0, 1, 2, 4], [0, 2]]
    y = df_current.iloc[[0, 1, 2, 4], 1]

    lr = LinearRegression()
    lr.fit(X, y)

    df_current.iloc[3, 1] = lr.predict(
        df_current.iloc[3, [0, 2]].values.reshape(1, 2)
    )[0]

    # ---- Marketing Spend (row 49) ----
    X = df_current.iloc[0:49, [0, 1]]
    y = df_current.iloc[0:49, 2]

    lr = LinearRegression()
    lr.fit(X, y)

    df_current.iloc[49, 2] = lr.predict(
        df_current.iloc[49, [0, 1]].values.reshape(1, 2)
    )[0]

    # Calculate change from previous iteration
    diff = np.abs(
        df_current.iloc[[1, 3, 49], :] -
        df_previous.iloc[[1, 3, 49], :]
    ).sum().sum()

    print(f"Iteration {iteration}: difference = {diff}")

    # Stop when changes are almost zero
    if diff < 1e-6:
        print("Converged!")
        break
```
