```python
import pandas as pd
import numpy as np

from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

```python
df = pd.read_csv("/Advertising Budget and Sales.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
df.head()
```

```python
# Sales=β0​+β1​(TV)+β2​(Radio)+β3​(Newspaper) in simple linear reg we have some thing y = Bo+B1X
# and the sheet we create is called regression plane not reg line in multiple linear reg

df.describe()
```

```python
x = df.iloc[:,0:3]
y = df.iloc[:,-1]
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(X_train,y_train)
```

```python
y_pred =lr.predict(X_test)
```

```python
y_pred
```

```python
print("MSE" ,mean_squared_error(y_test,y_pred))
print("MAE" ,mean_absolute_error(y_test,y_pred))
print("r2 Score" ,r2_score(y_test,y_pred))
print("RMSE" ,np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
# so Root mean squared error tell us that our model is 2.12 sales units away from the true value.
```

```python
import plotly.express as px
import plotly.graph_objects as go
```

```python
# Regression plane visualization for Tv ad column or radio one

# Create grid
x_tv = np.linspace(df['TV Ad Budget ($)'].min(), df['TV Ad Budget ($)'].max(), 30)
x_radio = np.linspace(df['Radio Ad Budget ($)'].min(), df['Radio Ad Budget ($)'].max(), 30)

xGrid_tv, xGrid_radio = np.meshgrid(x_tv, x_radio)

# Calculate the mean of 'Newspaper Ad Budget ($)' to keep it constant for the 3D plane visualization
mean_newspaper = df['Newspaper Ad Budget ($)'].mean()

# Create a constant array for 'Newspaper Ad Budget ($)' with the mean value
xGrid_newspaper = np.full(xGrid_tv.shape, mean_newspaper)

# Combine the grids into a single array for prediction
final = np.vstack((xGrid_tv.ravel(), xGrid_radio.ravel(), xGrid_newspaper.ravel())).T

z = lr.predict(final).reshape(30, 30)

# 3D Plot
fig = px.scatter_3d(
    df,
    x='TV Ad Budget ($)',
    y='Radio Ad Budget ($)',
    z='Sales ($)',
    opacity=0.8
)

fig.add_trace(
    go.Surface(
        x=x_tv,
        y=x_radio,
        z=z,
        opacity=0.5,
        showscale=False
    )
)

fig.show()
```

```python
# Regression plane visualization for Tv ad column or news one

#Create grid
x_tv = np.linspace(df['TV Ad Budget ($)'].min(), df['TV Ad Budget ($)'].max(), 30)
x_news = np.linspace(df['Newspaper Ad Budget ($)'].min(), df['Newspaper Ad Budget ($)'].max(), 30)

xGrid_tv, xGrid_news = np.meshgrid(x_tv, x_news)

# Calculate the mean of 'Newspaper Ad Budget ($)' to keep it constant for the 3D plane visualization
mean_radio = df['Radio Ad Budget ($)'].mean()

# Create a constant array for 'Newspaper Ad Budget ($)' with the mean value
xGrid_radio = np.full(xGrid_tv.shape, mean_radio)

# Combine the grids into a single array for prediction
final = np.vstack((xGrid_tv.ravel(), xGrid_news.ravel(), xGrid_radio.ravel())).T

z = lr.predict(final).reshape(30, 30)

# 3D Plot
fig = px.scatter_3d(
    df,
    x='TV Ad Budget ($)',
    y='Newspaper Ad Budget ($)',
    z='Sales ($)',
    opacity=0.8
)

fig.add_trace(
    go.Surface(
        x=x_tv,
        y=x_news,
        z=z,
        opacity=0.5,
        showscale=False
    )
)

fig.show()
```

```python
# Regression plane visualization for Radio + Newspaper

# Create grid
x_radio = np.linspace(
    df['Radio Ad Budget ($)'].min(),
    df['Radio Ad Budget ($)'].max(),
    30
)

x_news = np.linspace(
    df['Newspaper Ad Budget ($)'].min(),
    df['Newspaper Ad Budget ($)'].max(),
    30
)

xGrid_radio, xGrid_news = np.meshgrid(x_radio, x_news)

# Keep TV constant at its mean
mean_tv = df['TV Ad Budget ($)'].mean()

xGrid_tv = np.full(xGrid_radio.shape, mean_tv)

# Combine the grids for prediction
final = np.vstack((
    xGrid_tv.ravel(),
    xGrid_radio.ravel(),
    xGrid_news.ravel()
)).T

# Predict Sales
z = lr.predict(final).reshape(30, 30)

# 3D Plot
fig = px.scatter_3d(
    df,
    x='Radio Ad Budget ($)',
    y='Newspaper Ad Budget ($)',
    z='Sales ($)',
    opacity=0.8
)

fig.add_trace(
    go.Surface(
        x=x_radio,
        y=x_news,
        z=z,
        opacity=0.5,
        showscale=False
    )
)

fig.show()
```
