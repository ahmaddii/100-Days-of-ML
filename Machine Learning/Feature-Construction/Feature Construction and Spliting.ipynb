```python
import pandas as pd
import numpy as np

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("train2.csv")[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
x = df.iloc[:,0:4]
y = df.iloc[:,-1]
```

```python
y
```

```python
np.mean(cross_val_score(LogisticRegression(),x,y,scoring="accuracy",cv=20))
```

<h1>Feature Construction</h1>

```python
x["Family_size"] = x["SibSp"] + x["Parch"] + 1
```

```python
x.head()
```

```python
def family(num):
    if num == 0:
        return 0 #alone
    elif num > 1 and num <=4:
        return 1 # means small family
    else:
        return 2 # large family
```

```python
family(5)
```

```python
x["FamilyType"] = x["Family_size"].apply(family)
```

```python
x.sample(20)
```

```python
x.drop(columns=["SibSp","Parch","Family_size"],inplace=True)
```

```python
x.head(10)
```

```python
np.mean(cross_val_score(LogisticRegression(),x,y,scoring='accuracy',cv=20))
```
