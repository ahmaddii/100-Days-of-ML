```python
import numpy as np
import pandas as pd
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import scipy.stats as stats

from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

from sklearn.preprocessing import PowerTransformer
```

```python
df = pd.read_csv("concrete_data.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
x = df.drop(columns="Strength")
y = df.iloc[:,-1]
```

```python
X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
lr = LinearRegression()

lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
r2_score(y_test,y_pred) # why usse r2 score it use in regression problem means numerical target column and it dosent compare actual or predict it checks how close both values are
```

```python
lr = LinearRegression()

np.mean(cross_val_score(lr,x,y,scoring="r2",cv=10))
```

```python
# ploting the distplot and qq plot without transformation

for col in X_train.columns:
    plt.figure(figsize=(14,4))
    plt.subplot(121)
    sns.distplot(X_train[col])
    plt.title(col)

    plt.subplot(122)
    stats.probplot(X_train[col],dist="norm",plot=plt)
    plt.title(col)

    plt.show()
```

```python
# now apply box cox transformation

pt = PowerTransformer(method="box-cox")

X_train_transformed = pt.fit_transform(X_train+0.000001)
X_test_transformed = pt.transform(X_test+0.000001)

pd.DataFrame({"cols":X_train.columns,"box_cox_lambdas":pt.lambdas_}) # new cols with padnas datafarme and we generate lambda value for each column
```

```python
# (540)'0.153

lr = LinearRegression()

lr.fit(X_train_transformed,y_train)
```

```python
y_pred = lr.predict(X_test_transformed)
```

```python
r2_score(y_test,y_pred)
```

```python
# Using cross val score

pt = PowerTransformer(method='box-cox')
X_transformed = pt.fit_transform(x+0.0000001)

lr = LinearRegression()
np.mean(cross_val_score(lr,X_transformed,y,scoring='r2',cv=10))
```

```python
# ploting the distplot and qq plot with transformation

X_train_transformed = pd.DataFrame(X_train_transformed,columns=X_train.columns)

for col in X_train.columns:
    plt.figure(figsize=(14,4))
    plt.subplot(121)
    sns.distplot(X_train[col])
    plt.title(col)

    plt.subplot(122)
    sns.distplot(X_train_transformed[col])
    plt.title(col)

    plt.show()
```

```python
pt = PowerTransformer()

X_train_transformed = pt.fit_transform(X_train)
X_test_transformed = pt.fit_transform(X_test)

lr = LinearRegression()

lr.fit(X_train_transformed,y_train)
```

```python
y_pred = lr.predict(X_test_transformed)

r2_score(y_test,y_pred)
```
